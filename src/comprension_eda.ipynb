### Corregimos los outliers de edad

```python
df_clientes_banco = df_clientes_banco.loc[df_clientes_banco['edad_cliente'] <= 90]
```

### Tratamos los nulos

#### tendencia_ingresos

```python
#definimos una funcion para que tendencia_ingresos unifique los valores y elimine los nulos
def clasificar_tipo_tendencias(x):
    if isinstance(x, (int, float)):
        if pd.isna(x):
            return 'Sin Especificar'
        elif x > 0:
            return "Creciente"
        elif x < 0:
            return "Decreciente"
        else:
            return "Estable"
    return x

#aplicamos la funcion
df_clientes_banco['tendencia_ingresos'] = df_clientes_banco['tendencia_ingresos'].apply(clasificar_tipo_tendencias)
```

#### promedio_ingresos_datacredito

```python
df_clientes_banco['promedio_ingresos_datacredito'] = df_clientes_banco['promedio_ingresos_datacredito'].fillna('Sin Especificar')
```

#### saldo_mora_codeudor

```python
display(df_clientes_banco.groupby(df_clientes_banco['saldo_mora_codeudor'].isna())['puntaje'].describe())
```

```python
df_clientes_banco['saldo_mora_codeudor'] = df_clientes_banco['saldo_mora_codeudor'].fillna(0)
```

#### saldo_principal, saldo_mora y saldo_total

```python
#usamos una matriz de correlacion para verificar la relacion entre variables
df_clientes_banco[['saldo_principal', 'saldo_mora', 'saldo_total']].corr()
```

```python
#verificamos las otras dos variables de saldo cuando saldo_principal es nulo
df_clientes_banco[
    df_clientes_banco['saldo_principal'].isna()
][['saldo_mora', 'saldo_total']].describe()
```

```python
#creamos una mascara para filtrar por los nulos de saldo_principal que a su vez contenga 0 en las otras dos variables
mask = (
    df_clientes_banco['saldo_principal'].isna() &
    (df_clientes_banco['saldo_mora'] == 0) &
    (df_clientes_banco['saldo_total'] == 0)
)

#imputo con 0 en los casos de la mascara
df_clientes_banco.loc[mask, 'saldo_principal'] = 0
```

```python
#dispongo los casos en los que se registra el pago a tiempo pero se registra deuda
df_clientes_banco[
    (df_clientes_banco['Pago_atiempo'] == 1) &
    (df_clientes_banco['saldo_mora'] > 0)
]
```

```python
#imputo con 0 los nulos restantes de los saldos
df_clientes_banco[['saldo_principal', 'saldo_mora', 'saldo_total']] = (
    df_clientes_banco[['saldo_principal', 'saldo_mora', 'saldo_total']].fillna(0)
)
```

#### puntaje_datacredito

```python
#veridico los casos en los que la variable es nula
df_clientes_banco[
    df_clientes_banco['puntaje_datacredito'].isna()
]
```

```python
#verificamos que no queden valores faltantes
display(df_clientes_banco.isnull().sum().sort_values(ascending=False))
```

### Analisis bivariable

```python
plt.figure(figsize=(8,4))

sns.violinplot(
    data=df_salarios,
    x='Pago_atiempo',
    y='salario_cliente'
)

plt.title('Salario según pago a tiempo')
plt.xlabel('Pago a tiempo')
plt.ylabel('Salario')

plt.ticklabel_format(style='plain', axis='y')

plt.show()
```

La mayoria de los clientes, independientemente de si pagan o no tienen salarios de entre 1 y 4 millones, sin embargo existen unos pocos clientes con salarios extremadamente altos. El salario casi no tiene ingerencia en si el cliente paga o no.

```python
plt.figure(figsize=(8,4))

sns.countplot(
    data=df_clientes_banco,
    x='cant_creditosvigentes',
    hue='Pago_atiempo'
)

handles, _ = plt.gca().get_legend_handles_labels()

plt.legend(
    handles=handles,
    labels=['No', 'Sí'],
    title='Pago a tiempo'
)

plt.title('Cantidad de créditos vigentes según pago a tiempo')
plt.xlabel('Créditos vigentes')
plt.ylabel('Cantidad de clientes')
plt.xticks(rotation=45)

plt.show()
```

Se observa como la distribucion sesgada a la derecha indica que la mayoria de los clientes tienen pocos creditos con un pico en los 4 creditos vigentes, sin embargo se puede observar tambien como los creditos impagos tienen dos picos, uno a los 4 creditos vigentes y otro a los 8. Esto nos puede dar a entender que la mayor cantidad de creditos vigentes son adquiridos por casos muy aislados (alejados de la moda de 4 creditos) o usuarios muy especificos.

```python
tabla_creditos = pd.crosstab(
    df_clientes_banco['tipo_credito'],
    df_clientes_banco['Pago_atiempo'],
    normalize='index'
)

tabla_creditos.plot(
    kind='bar',
    stacked=True,
    figsize=(8,4)
)

plt.legend(
    labels=['No', 'Si'],
    title='Pago a tiempo'
)

plt.title('Pago a tiempo según tipo de crédito')
plt.ylabel('Proporción')
plt.xlabel('Tipo crédito')

plt.show()
```

El tipo de credito no tiene en si mucha ingerencia en el pago pero si se remarcan diferencias como que los usuarios que adquieren el tipo de credito 6 son menos propensos a pagar a tiempo.

```python
plt.figure(figsize=(8,4))

sns.violinplot(
    data=df_clientes_banco,
    x='Pago_atiempo',
    y='plazo_meses'
)

plt.title('Plazo del crédito según pago a tiempo')
plt.ylabel('Plazo en meses')
plt.xlabel('Pago a tiempo')
plt.show()
```

A mayor plazo de cuota mas propenso es el usuario de pagar como se debe, si bien son los pocos los que pactan los plazos mas altos. Los usuarios que pagan a tiempo se distribuyen de manera mas erratica.

```python
pd.crosstab(
    df_clientes_banco['tendencia_ingresos'],
    df_clientes_banco['Pago_atiempo'],
    normalize='index'
)

sns.countplot(
    data=df_clientes_banco,
    x='tendencia_ingresos',
    hue='Pago_atiempo'
)

plt.legend(
    labels=['No','Si'],
    title='Pago a tiempo'
)

plt.title('Tendencia ingresos vs pago a tiempo')
plt.xlabel('Tendencia de ingresos')
plt.ylabel('Cantidad')
plt.show()
```

Las personas que tienen tendencia de ingresos creciente son los que mas se atrasan en los pagos, de todas formas se puede observar como la tendencia de ingresos es mas influyente en el pago a tiempo.

```python
plt.figure(figsize=(10,5))

sns.countplot(
    data=df_clientes_banco,
    y='tipo_laboral',
    hue='Pago_atiempo'
)
plt.legend(
    labels=['No','Si'],
    title='Pago a tiempo'
)

plt.title('Tipo laboral según pago a tiempo')
plt.xlabel('Cantidad')
plt.ylabel('Tipo laboral')
plt.show()
```

Aunque muy pequeña la diferencia se observa como los clientes empleados son los que mas se atrasan en sus pagos, sin embargo se puede deber a que los clientes empleados son la mayoria dado tambien por su alta cantidad de pagos a tiempo.

```python
df_clientes_banco.groupby('Pago_atiempo')[
    [
        'salario_cliente',
        'total_otros_prestamos',
        'puntaje',
        'puntaje_datacredito'
    ]
].mean()
```

Se disponen las medias de las variables de salario, otros prestamos, el puntaje y el puntaje de datacredito para observar que numeros maneja la mayoria de la poblacion de clientes. Se puede ver como el puntaje es realmente influyente en si pagan o no, sin embargo el puntaje de datacredito disminuye muy poco entre quienes pagan y quienes no. Tambien se puede observar como la diferencia de salario entre quienes pagan y quienes no es pequeña aunque no diminuta.

### Analisis multivariado

```python
df_numerico = df_clientes_banco.select_dtypes(include='number').copy()
df_numerico['Pago_atiempo'] = df_clientes_banco['Pago_atiempo'].astype(int)

correlacion = df_numerico.corr()

plt.figure(figsize=(14,10))

sns.heatmap(
    correlacion,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    linewidths=0.5
)

plt.title('Matriz de correlación')
plt.show()
```

Se puede ver como no hay demasiada correlatividad, pero hay variables que denotan algunas cuestiones como que creditos del sector financiero tiene una relacion de 0.79 con cantidad de creditos vigentes por lo que se entiende que la mayoria de creditos solicitados son de dicho sector, o como que al aumentar la cuota pactada aumenta el capital prestado, o como que el saldo principal representa la mayoria del saldo total. La unica variable numerica que es directamente influyente en el pago a tiempo es el puntaje, lo cual se entiende ya que el puntaje nace a partir de si el credito se paga en tiempo o no.
